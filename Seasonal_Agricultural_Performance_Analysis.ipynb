# Data Cleaning and Data preparation

```python
# Remove exact duplicate rows
df = df.drop_duplicates().copy()

# Check Farm_ID uniqueness
duplicate_farm_ids = df["Farm_ID"].duplicated().sum()
print("Duplicate Farm_ID values:", duplicate_farm_ids)

# Convert categorical fields to string/category-friendly values
categorical_columns = ["State", "District", "Crop", "Season", "Irrigation_Method"]

for col in categorical_columns:
    df[col] = df[col].astype("string").str.strip()

# Impute environmental missing values using the median
for col in ["Rainfall_mm", "Soil_Moisture_pct"]:
    df[col] = df[col].fillna(df[col].median())

# Yield is important for performance analysis.
# Impute missing yield using Crop + Season median where possible,
# then use the overall median as a fallback.

df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(
    df.groupby(["Crop", "Season"])["Yield_Tonnes_Ha"].transform("median")
)
df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(df["Yield_Tonnes_Ha"].median())

print("Remaining missing values:")
print(df.isna().sum()[df.isna().sum() > 0])
print("\nCleaned dataset shape:", df.shape)
```

# Seasonal Agricultural Performance

```python
season_summary = df.groupby("Season").agg(
    Farms=("Farm_ID", "count"),
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Median_Yield=("Yield_Tonnes_Ha", "median"),
    Average_Production=("Production_Tonnes", "mean"),
    Total_Production=("Production_Tonnes", "sum"),
    Average_Profit=("Profit_INR", "mean"),
    Total_Profit=("Profit_INR", "sum"),
    Average_Market_Price=("Market_Price_INR_Tonne", "mean"),
    Average_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean"),
    Average_Rainfall=("Rainfall_mm", "mean"),
    Average_Temperature=("Avg_Temperature_C", "mean"),
    Average_Disease_Risk=("Disease_Pest_Risk_pct", "mean")
).sort_values("Average_Yield", ascending=False)

print(season_summary.round(2))
```

#Chart 2 - Average Yield by Season

```python
plt.figure(figsize=(8, 5))
plt.bar(
    season_summary.index,
    season_summary["Average_Yield"]
)
plt.title("Average Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()

best_yield_season = season_summary["Average_Yield"].idxmax()
lowest_yield_season = season_summary["Average_Yield"].idxmin()

print("Observation:")
print(f"{best_yield_season} has the highest average yield at "
      f"{season_summary.loc[best_yield_season, 'Average_Yield']:.2f} tonnes/ha.")
print(f"{lowest_yield_season} has the lowest average yield at "
      f"{season_summary.loc[lowest_yield_season, 'Average_Yield']:.2f} tonnes/ha.")
```

# Outlier Analysis Using IQR

```python
outlier_results = []

for col in numeric_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_results.append([
        col, q1, q3, lower, upper, int(count)
    ])

outlier_table = pd.DataFrame(
    outlier_results,
    columns=["Variable", "Q1", "Q3", "Lower_Bound", "Upper_Bound", "Outlier_Count"]
)

outlier_table["Outlier_Percentage"] = (
    outlier_table["Outlier_Count"] / len(df) * 100
)

print(outlier_table.sort_values("Outlier_Count", ascending=False).round(2))
```
